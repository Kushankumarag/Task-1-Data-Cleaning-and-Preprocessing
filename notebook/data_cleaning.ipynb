**Internship:** Data Analyst Internship
**Tools:** Python, Pandas, Jupyter Notebook

```python
import pandas as pd
print('Libraries imported successfully.')
```

## Step 1: Load the Dataset

```python
df = pd.read_csv('../dataset/netflix_titles.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

## Step 2: Check Missing Values

```python
df.isnull().sum()
```

## Step 3: Standardize Column Names

```python
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(' ', '_', regex=False)
    .str.replace(r'[^\w]', '', regex=True)
)
print(list(df.columns))
```

## Step 4: Handle Missing Values

```python
df.fillna({
    'director': 'Unknown',
    'cast': 'Unknown',
    'country': 'Unknown',
    'rating': 'Not Rated',
    'duration': 'Unknown'
}, inplace=True)
df.isnull().sum()
```

## Step 5: Convert date_added to Datetime

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
df['date_added'].dtype
```

## Step 6: Check and Remove Duplicates

```python
duplicate_count = df.duplicated().sum()
print(f'Duplicate rows found: {duplicate_count}')
df = df.drop_duplicates()
print(f'Final shape: {df.shape}')
```

## Step 7: Verify Data Types

```python
df.dtypes
```

## Step 8: Export Cleaned Dataset

```python
df.to_csv('../dataset/cleaned_netflix_titles.csv', index=False)
print('Cleaned dataset saved successfully.')
```
