**Dataset:** Netflix Movies and TV Shows  
**Tools:** Python, Pandas, Jupyter Notebook

```python
import pandas as pd
print('Libraries imported successfully.')
```

## Step 1: Load the Dataset

```python
df = pd.read_csv('../dataset/netflix_titles.csv')
print(f'Shape: {df.shape}')
df.head()
```

## Step 2: Inspect the Dataset

```python
df.info()
```

## Step 3: Check Missing Values

```python
print('Missing values per column:')
df.isnull().sum()
```

## Step 4: Standardize Column Names

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_', regex=False)
print('Standardized columns:')
print(list(df.columns))
```

## Step 5: Handle Missing Values

```python
fill_values = {
    'director': 'Unknown',
    'cast': 'Unknown',
    'country': 'Unknown',
    'rating': 'Not Rated',
    'duration': 'Unknown'
}
df.fillna(fill_values, inplace=True)
print('Missing values after filling:')
df.isnull().sum()
```

## Step 6: Convert date_added to Datetime

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
print(f'date_added dtype: {df["date_added"].dtype}')
```

## Step 7: Check and Remove Duplicates

```python
duplicate_count = df.duplicated().sum()
print(f'Duplicate rows found: {duplicate_count}')
df = df.drop_duplicates()
print(f'Final shape: {df.shape}')
```

## Step 8: Verify Data Types

```python
df.dtypes
```

## Step 9: Export Cleaned Dataset

```python
df.to_csv('../dataset/cleaned_netflix_titles.csv', index=False)
print('Cleaned dataset saved to dataset/cleaned_netflix_titles.csv')
```
